# Plotting an Electron Density Graph

In this tutorial we will plot an SCF electron density using the information about the basis set we used and the self-consistent density matrix $\textbf{D}$.

## Import Section

We will need Psi4 to do an SCF calculation, NumPy for the matrix and array manipulations, and matplotlib to make the plots.

```python
import psi4
import numpy as np
%matplotlib inline
import matplotlib.pyplot as plt 
```

## SCF Calculation of the Electron Density

Let us do the SCF calculation for the Ne atom at the RHF/STO-3G level of theory (later, we will try other basis sets too):

```python
psi4.core.set_output_file('output.dat', False)

mol = psi4.geometry("""
0 1 
Ne
""")

psi4.set_options({'basis': 'STO-3G',
                  'reference': 'RHF'})

scf_e, scf_wfn = psi4.energy('scf', return_wfn=True)
print("Psi4 SCF Energy is: %.6f Hartree" % scf_e)
```

Now, we also can print the self-consistent density matrix (since the calculation is spin-restricted, $\textbf{D}_{\alpha} = \textbf{D}_{\beta}$):

```python
D = scf_wfn.Da().nph[0]
print(D)
```

## Basis Set Expansion

The elecron density for a single-reference methods (using a single Slater determinant, i.e., HF and KS-DFT) is given by: $$\rho(\textbf{r})=\sum_{i=1}^{\text{occ}}|\varphi_i(\textbf{r})|^2.$$ Where $\varphi_i(\textbf{r})$ is a molecular orbital (MO) expressed as a linear combination of basis functions of the basis set we have chosen: $$\varphi_i(\textbf{r}) = \sum_{\mu=1}c_{\mu i}\phi_{\mu}(\textbf{r}).$$ where the MO coefficients $c_{\mu i}$ are obtained via the SCF procedure and constitute the MO coefficients matrix $\textbf{C}$. These matrix is a building block of the density matrix $\textbf{D} = \textbf{C}^{\dagger}\textbf{C}$.

Since we are operating within the Gaussian-type orbital (GTO) basis set, each basis function itself a linear combination of the GTO functions, also known as primitive gaussians: $$\phi_{\mu}(\textbf{r}) = \sum_{p=1}d_{p\mu}g_{p}(\alpha_{p\mu},\textbf{r}),\quad g_{p}(\alpha_{p\mu},\textbf{r})=\left(\frac{2\alpha_{p\mu}}{\pi}\right)^{3/4}e^{-\alpha_{p\mu}r^2}.$$
