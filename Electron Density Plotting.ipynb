# Plotting an Electron Density Graph

In this tutorial we will plot an SCF electron density using the information about the basis set we used and the self-consistent density matrix $\textbf{D}$.

## Import Section

We will need Psi4 to do an SCF calculation, NumPy for the matrix and array manipulations, and matplotlib to make the plots.

```python
import psi4
import numpy as np
%matplotlib inline
import matplotlib.pyplot as plt 
```

## SCF Calculation of the Electron Density

Let us do the SCF calculation for the Ne atom at the RHF/STO-3G level of theory (later, we will try other basis sets too):

```python
psi4.core.set_output_file('output.dat', False)

mol = psi4.geometry("""
0 1 
Ne
""")

psi4.set_options({'basis': 'STO-3G',
                  'reference': 'RHF'})

scf_e, scf_wfn = psi4.energy('scf', return_wfn=True)
print("Psi4 SCF Energy is: %.6f Hartree" % scf_e)
```

Now, we also can print the self-consistent density matrix (since the calculation is spin-restricted, $\textbf{D}_{\alpha} = \textbf{D}_{\beta}$):

```python
D = scf_wfn.Da().nph[0]
print(D)
```
