```python
import sys
import psi4
import numpy as np

%load_ext autoreload
%autoreload 2
sys.path.append('/Users/ivanbosko/Documents/CODES/GIT/DFT-Tools/SCF') 

import Kohn_Sham.RKS as RKS
import Kohn_Sham.UKS as UKS

import Hartree_Fock.RHF as RHF
```

```python
import custom_functionals
EXX = custom_functionals.EXX
LDAx = custom_functionals.LDAx
h2o = psi4.geometry(f"""
    units bohr
    0 1
    O
    H 1 1.81
    H 1 1.81 2 104.5                      
    symmetry c2v
    """)
ne = psi4.geometry(f"""
    units bohr
    0 1
    Ne                  
    symmetry c1
    """)
h = psi4.geometry(f"""
    units bohr
    0 2
    H                  
    symmetry c1
    """)

psi4.core.set_output_file('output.dat', False)

# method = EXX
method = LDAx

mol = h2o
# mol = ne
# mol = h

psi4.set_options({"basis": "UGBS",
                  "DFT_SPHERICAL_POINTS": 302,
                  "DFT_RADIAL_POINTS": 350
                  })
```

```python
E, D, homo, scf_iter = RKS.ks_solver(mol, method)
print(f"\nFinal SCF Energy: {E:.6f} Hartree")
```

```python
E, Da, Db, scf_iter = UKS.ks_solver(mol, method)
print(f"\nFinal SCF Energy: {E:.6f} Hartree")
```

```python
psi4.set_options({"scf_type": "PK",
                  "d_convergence": 1e-8,
                  "e_convergence": 1e-8
                  })
psi4_E = psi4.energy('SCF', dft_functional=method, molecule=mol, return_wfn=False)
print(f"Psi4 SCF Energy: {psi4_E:.6f} Hartree")
```

```python
E, D, homo, scf_iter = RHF.hf_solver(mol, verbose=True)
print(f"\nFinal SCF Energy: {E:.6f} Hartree")
```

```python
psi4.set_options({"scf_type": "PK",
                  "d_convergence": 1e-8,
                  "e_convergence": 1e-8,
                  "GUESS": "Core"
                  })
psi4_E = psi4.energy('SCF', molecule=mol, return_wfn=False)
print(f"Psi4 SCF Energy: {psi4_E:.6f} Hartree")
```
