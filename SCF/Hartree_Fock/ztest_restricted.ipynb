```python
import sys
import psi4
import numpy as np

%load_ext autoreload
%autoreload 2
sys.path.append('/Users/ivanbosko/Documents/CODES/GIT/DFT-Tools/SCF') 

import Hartree_Fock.RHF as RHF
```

```python
h2o = psi4.geometry(f"""
    units bohr
    0 1
    O
    H 1 1.81
    H 1 1.81 2 104.5                      
    symmetry c2v
    """)
ne = psi4.geometry(f"""
    units bohr
    0 1
    Ne                  
    symmetry c1
    """)
h = psi4.geometry(f"""
    units bohr
    0 2
    H                  
    symmetry c1
    """)

psi4.core.set_output_file('output.dat', False)

mol = h2o
# mol = ne
# mol = h

## Four last settings are exclusively for Psi4 program
## to work in the same mode as the custom RHF.py
psi4.set_options({"basis": "UGBS",
                  "scf_type": "PK",
                  "d_convergence": 1e-8,
                  "e_convergence": 1e-8,
                  "GUESS": "Core"
                  })
```

```python
%%time
E, D, homo, scf_iter = RHF.scf_solver(mol, verbose=True)
print(f"\nFinal SCF Energy: {E:.6f} Hartree")
```

```python
%%time
psi4_E = psi4.energy('SCF', molecule=mol, return_wfn=False)
print(f"Psi4 SCF Energy: {psi4_E:.6f} Hartree")
```
